```python

import pandas as pd

data = {
    "Student_Name": [
        "Amit", "Priya", "Rahul", "Sneha", "Neha",
        "Rohan", "Pooja", "Akash", "Kiran", "Sahil"
    ],
    "Maths": [78, 90, 45, 88, 67, 32, 95, 56, 72, 60],
    "Science": [85, 92, 50, 79, 72, 40, 91, 60, 80, 65],
    "English": [70, 86, 55, 90, 65, 35, 88, 48, 75, 58]
}

df = pd.DataFrame(data)

df
```

```python

# Check the number of rows and columns
print("Dataset Shape:", df.shape)

# Check column names and data types
print("\nDataset Information:")
df.info()

# Display first 5 rows
print("\nFirst 5 Students:")
print(df.head())
```

```python

print("Missing Values:")
print(df.isnull().sum())
```

```python

print("Duplicate Rows:", df.duplicated().sum())
```

```python

print(df[["Maths", "Science", "English"]].describe())
```

```python

subjects = ["Maths", "Science", "English"]

df["Total_Marks"] = df[subjects].sum(axis=1)

df[["Student_Name", "Maths", "Science", "English", "Total_Marks"]]
```

```python

# Step 1: Calculate total marks
subjects = ["Maths", "Science", "English"]

df["Total_Marks"] = df[subjects].sum(axis=1)

# Step 2: Calculate average marks
df["Average_Marks"] = df[subjects].mean(axis=1).round(2)

# Step 3: Calculate Pass/Fail
df["Result"] = df[subjects].ge(35).all(axis=1).map({
    True: "Pass",
    False: "Fail"
})

# Step 4: Display complete dataset
display(df)
```

```python
print(df.columns.tolist())
```

```python

print("Total Students:", len(df))
print("Class Average:", round(df["Average_Marks"].mean(), 2))

print("\nPass/Fail Summary:")
print(df["Result"].value_counts())
```

```python

topper = df.loc[df["Total_Marks"].idxmax()]

print("Topper:", topper["Student_Name"])
print("Total Marks:", topper["Total_Marks"])
print("Average Marks:", topper["Average_Marks"])
```

```python

subjects = ["Maths", "Science", "English"]

print(df[subjects].mean().round(2))
```

```python

import matplotlib.pyplot as plt

subjects = ["Maths", "Science", "English"]
averages = df[subjects].mean()

plt.figure(figsize=(7, 4))
averages.plot(kind="bar")

plt.title("Subject-wise Average Marks")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(10, 5))

plt.bar(df["Student_Name"], df["Average_Marks"])

plt.title("Student Performance Analysis")
plt.xlabel("Student Name")
plt.ylabel("Average Marks")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

result_counts = df["Result"].value_counts()

plt.figure(figsize=(5, 5))
result_counts.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Pass vs Fail Percentage")
plt.ylabel("")
plt.show()
```

```python

# 1. Find the topper
topper = df.loc[df["Total_Marks"].idxmax()]

print("STUDENT PERFORMANCE ANALYSIS")
print("-" * 35)

print("Total Students:", len(df))
print("Class Average:", round(df["Average_Marks"].mean(), 2))

print("\nTop Performer:", topper["Student_Name"])
print("Topper's Total Marks:", topper["Total_Marks"])

# 2. Subject-wise averages
subjects = ["Maths", "Science", "English"]
subject_avg = df[subjects].mean()

print("\nSubject-wise Average Marks:")
print(subject_avg.round(2))

print("\nHighest Average Subject:", subject_avg.idxmax())
print("Lowest Average Subject:", subject_avg.idxmin())

# 3. Pass percentage
passed = (df["Result"] == "Pass").sum()
failed = (df["Result"] == "Fail").sum()

print("\nStudents Passed:", passed)
print("Students Failed:", failed)
print("Pass Percentage:", round(passed / len(df) * 100, 2), "%")
```
